## The supervised ML pipeline

<span style="background-color: #FFFF00">**0. Data collection/manipulation**: you might have multiple data sources and/or you might have more data than you need</span>
   - you need to be able to read in datasets from various sources (like CSV, Excel, SQL, parquet, etc)
   - you need to be able to filter the columns/rows you need for your ML model
   - you need to be able to combine the datasets into one DataFrame 

**1. Exploratory Data Analysis (EDA)**: you need to understand your data and verify that it doesn't contain errors
   - do as much EDA as you can!
    
**2. Split the data into different sets**: most often the sets are train, validation, and test (or holdout)
   - practitioners often make errors in this step!
   - you can split the data randomly, based on groups, based on time, or any other non-standard way if necessary to answer your ML question

**3. Preprocess the data**: ML models only work if X and Y are numbers! Some ML models additionally require each feature to have 0 mean and 1 standard deviation (standardized features)
   - often the original features you get contain strings (for example a gender feature would contain 'male', 'female', 'non-binary', 'unknown') which needs to be transformed into numbers
   - often the features are not standardized (e.g., age is between 0 and 100) but it needs to be standardized
    
**4. Choose an evaluation metric**: depends on the priorities of the stakeholders
   - often requires quite a bit of thinking and ethical considerations
     
**5. Choose one or more ML techniques**: it is highly recommended that you try multiple models
   - start with simple models like linear or logistic regression
   - try also more complex models like nearest neighbors, support vector machines, random forest, etc.
    
**6. Tune the hyperparameters of your ML models (aka cross-validation or hyperparameter tuning)**
   - ML techniques have hyperparameters that you need to optimize to achieve best performance
   - for each ML model, decide which parameters to tune and what values to try
   - loop through each parameter combination
       - train one model for each parameter combination
       - evaluate how well the model performs on the validation set
   - take the parameter combo that gives the best validation score
   - evaluate that model on the test set to report how well the model is expected to perform on previously unseen data
    
**7. Interpret your model**: black boxes are often not useful
   - check if your model uses features that make sense (excellent tool for debugging)
   - often model predictions are not enough, you need to be able to explain how the model arrived to a particular prediction (e.g., in health care)

### Visual roadmap for the supervised ML pipeline

Today's lecture focuses on **Step 0**, which prepares raw data so the later ML steps are possible.

```text
Raw data sources
      |
      v
[0] Data collection and manipulation  <-- today's focus
      |
      v
[1] EDA -> [2] Split -> [3] Preprocess -> [4] Metric
      |
      v
[5] Choose models -> [6] Tune hyperparameters -> [7] Interpret model
```

**What to notice:** errors made in Step 0 can silently affect every later step. Before modeling, always check that the right files, rows, columns, and variable types were loaded.

# Data selection and collection issues
- The field is called **DATA** science for a reason!
    - Data is the most important part of data science.
    - The quality and quantity of data determine whether a project is feasible.
    - Data is usually much more valuable than algorithms.
    - Working with data takes up ~80% of a data scientist's time.
- When you start working on a new project, approach your dataset with healthy skepticism!
- Ask questions in two main categories:
    - Is the data appropriate for the problem you are trying to solve?
    - Is the data accurate?


## Is the dataset appropriate for the purpose?
- Can you answer your question with your data?
    - medical studies based on white men only, can you use it to diagnose everyone?
    - sometimes the answer is no! some diseases can affect men and women differently or various racial groups
    - sometimes the answer is yes! some diseases manifest the same way in everyone
- Is your dataset timely?
    - goal: predict COVID cases today
        - COVID changed a lot over the years: spreads easier but symptoms are milder
        - COVID data from 2020 and 2021 might not be usable today
    - goal: predict severe weather events like hurricanes
        - there are not many hurricanes in each year so the temptation is to use data going back as far as possible
        - hurricanes became more severe and more frequent in recent years
        - is it OK to use data from e.g., 1960s to predict hurricanes today?
- What biases are there in your dataset?
    - your ML model will learn any biases your data has
    - gender bias and racial bias are the main things to worry about when dealing with human data
- Is your dataset legal, ethical, and reliable?
    - can you use the dataset legally?
        - protected attributes (such as gender or race) often cannot be used especially in finance for example
    - ethical usage
        - if data is collected for one purpose, can you use it to solve another problem?
    - reliability
        - are there any conflicts of interests that might make the dataset unreliable?
        - example: climate data from big oil companies 

## Is the dataset accurate?
- Typos and errors
    - Mistakes by humans inputting data are extremely common!
- Why are there missing values in the dataset?
    - They could be fine.
        - Some respondents did not answer all the survey questions.
        - A doctor did not perform a test on all patients.
    - They could be due to instrument malfunction.
    - They could be due to changes in the data collection process over time.
- How are the missing values represented?
    - Sometimes as np.nan
    - Sometimes as a string like "missing" or "?"
    - Sometimes as unreasonable values
- Are the values valid?
    - Sometimes you will see incorrect or impossible values.
    - 6-digit ZIP codes
    - People older than 200 years
    - Negative numbers for a quantity that can only be non-negative
- Duplicate records
    - These could be due to data entry errors or data manipulation errors
      such as an incorrect merge or append.


# Documentation
- Document your dataset!
- Your future self and anyone else trying to reproduce your work will thank you!
- Documentation can be as simple as a text file.
- Describe each column in your dataset:
    - What is described in the column?
    - What quantity is measured? Does it have a unit?
    - What is the range of valid values?
    - What possible categories could there be? What does each category mean?
    - Are there missing values? If there are, why?
    - How are the missing values represented?
- You will see examples of this already today!
- **If the dataset for your final project does not come with documentation,
  you need to write it!**

## Pandas and Polars - why should you care?

- **When you work on an ML problem, you might work with data from various sources.**
    - Healthcare data might come from hospitals, insurance companies, state/federal agencies, etc.
    - Finance data could come from banks, brokerage accounts, Social Security offices, etc.
    - **You will need to pull data from all of these different sources and create one combined dataset ready for ML.**
- **You might also have more data than you need to solve the ML problem.**
    - In healthcare, you might be interested in people who have a certain symptom, or maybe you are interested in people who visited the ER multiple times.
    - In finance, you might be required by law to avoid sensitive or protected attributes, even if you have access to them.
    - **You need to filter out the rows and columns you need.**
- Packages like pandas and Polars make these tasks easier.


## Pandas and Polars intro

Similarities:
- Both are packages used for data manipulation and analysis.
- Both use the concept of DataFrames and Series. We will talk about this more later today.
- Both support reading and writing data in various formats, such as CSV, Excel, SQL, JSON, etc.
- The syntax is similar. Polars uses syntax similar to pandas to make it easier for pandas users to switch over.

Differences:
- pandas is more established. It was released in 2008, so it has a large user base, extensive documentation, and a large community.
- Polars is newer. It has only been around since 2020, but the user base is rapidly growing.
- pandas integrates with more packages than Polars, although that is changing.
- Polars is often much faster than pandas on large datasets.

When to use pandas?
- Most companies have extensive code bases in pandas, and it is unlikely they will switch over quickly because migrating code can be costly.
- If you work with small to medium datasets and computational speed and memory efficiency are not critical, pandas is fine.

When to use Polars?
- If you work on large datasets.
- If computational speed and memory efficiency are mission-critical.

Keep in mind that you can use both packages to make the best of both worlds:
- Use Polars to perform heavy computations.
- Use `pl.DataFrame.to_pandas()` and `pl.from_pandas()` to convert DataFrames as needed.

Check out [this](https://docs.pola.rs/user-guide/migration/pandas/) site to see how you can convert pandas code to Polars.

Check out [this](https://docs.pola.rs/user-guide/ecosystem/) site for a list of libraries and tools that support Polars.

### Packages

A package is a collection of classes and functions.
- a DataFrame (pd.DataFrame()) is a pandas class
    - a class is the blueprint of how the data should be organized 
    - classes have methods which can perform operations on the data (e.g., .head(), .shape)
- df is an object, an instance of the class.
    - when we put data into a class, it becomes an object 
    - methods are attached to objects 
       - you cannot call pd.head(), you can only call df.head()
- read_csv is a function
    - functions are called from the package
    - you cannot call df.read_csv, you can only call pd.read_csv()


### DataFrame structure: both rows and columns are indexed!
- index column, no name
    - contains the row names
    - by default, index is a range object from 0 to number of rows - 1 
    - any column can be turned into an index, so indices can be non-numeric, and also non-unique. More on this later.
- polars DataFrames do not have an index column by default! rows are indexed by their integer position in the table
    - you can add an index column if you'd like though 
- columns with column names on top in both polars and pandas

### Always print your DataFrame to check whether it looks good!

### Most common reasons it might not look OK:

   - The first row is not the column name.
        - There are rows above the column names that need to be skipped.
        - There is no column name, but by default, pandas assumes that the
          first row is the column name. As a result, the values of the first
          row end up as column names.
   - Character encoding is off.
   - The separator is not a comma but some other character.

## Filter rows in pandas
- Let's assume you have one DataFrame to work with, but you have too much data
  and you need to filter out some rows.
- There are several ways to do that:
##### 1) Integer-based indexing, the same way NumPy arrays are indexed
##### 2) Select rows based on the value of the index column
##### 3) Select rows based on column conditions


### `.iloc` versus `.loc` at a glance

| Selector | Uses | Slice stop included? | Example | Meaning |
| --- | --- | --- | --- | --- |
| `.iloc` | Integer position | No | `df.iloc[3:7]` | Rows in positions 3, 4, 5, and 6 |
| `.loc` | Index label | Yes, for label slices | `df.loc[3:7]` | Rows with index labels 3 through 7 |

**What to notice:** `.iloc` answers “where is the row located?” while `.loc` answers “what is the row's index label?” These are the same only when the index labels match the row positions.


### 1) Integer-based indexing, NumPy arrays are indexed the same way.


````python
# df_pd.iloc[]
# For more information, see:
# https://pandas.pydata.org/pandas-docs/stable/user_guide/indexing.html#indexing-integer
# iloc is how NumPy arrays are indexed (non-standard Python indexing).

# [start:stop:step] is the general indexing format.

# start, stop, and step are optional.
# print(df_pd)
# print(df_pd.iloc[:])
# print(df_pd.iloc[::])
# print(df_pd.iloc[::1])

# Select one row with 0-based indexing.
# print(df_pd.iloc[0])

# Index from the end of the DataFrame.
# print(df_pd.iloc[-2])
````

````python
# Select a slice. The stop index is not included.
# print(df_pd.iloc[3:7])

# Select every second element of the slice. The stop index is not included.
# print(df_pd.iloc[3:7:2])

# print(df_pd.iloc[3:7:-2])  # Return an empty DataFrame.
# print(df_pd.iloc[7:3:-2])
# Return rows with indices 7 and 5. 3 is the stop, so it is not included.

# This can be used to reverse rows.
# print(df_pd.iloc[::-1])

# Here is where indexing gets non-standard for Python.
# Select the 2nd, 5th, and 10th rows.
print(df_pd.iloc[[1, 4, 9]])
# This indexing does not work with lists, but it works with NumPy arrays.
````

**What to notice after the `.iloc` examples:**
- The stop value is excluded, just like standard Python slicing.
- A list of positions, such as `[1, 4, 9]`, selects specific rows by integer location.
- Negative positions count from the end of the DataFrame.



### 2) Select rows based on the value of the index column

````python
# df_pd.loc[]
# For more information, see:
# https://pandas.pydata.org/pandas-docs/stable/user_guide/indexing.html#indexing-label

# print(df_pd.index)
# The default index when reading in a file is a RangeIndex.
# In this case, .loc and .iloc work almost the same.

# One difference:
print(df_pd.loc[3:9:2])
# This selects the 4th, 6th, 8th, and 10th rows.
# The stop element is included!

# help(df_pd.set_index)
````

````python
df_index_age = df_pd.set_index("age", drop=False)

# print(df_pd.head())
# print(df_index_age.head())
# print(df_index_age.index)
# print(df_index_age.head())

# Collect everyone with age 30. The index is non-unique.
print(df_index_age.loc[30].head())
````

**What to notice after the `.loc` examples:**
- With label-based slicing, the stop label is included.
- After `set_index("age")`, the index is no longer unique, so `.loc[30]` can return many rows.
- A non-unique index can be useful, but it can also surprise you if you expected one row per index value.


### 3) Select rows based on column conditions

### Column-condition filters: pandas and Polars side by side

| Goal | pandas pattern | Polars pattern |
| --- | --- | --- |
| One condition | `df_pd[df_pd["age"] == 30]` | `df_pl.filter(df_pl["age"] == 30)` |
| Two conditions with AND | `df_pd[(cond1) & (cond2)]` | `df_pl.filter((cond1) & (cond2))` |
| Two conditions with OR | `df_pd[(cond1) \| (cond2)]` | `df_pl.filter((cond1) \| (cond2))` |

**What to notice:** each condition must be wrapped in parentheses before combining conditions with `&` or `|`.


````python
# One condition.
# print(df_pd[df_pd["age"] == 30].head())

# Here is the condition: it is a Boolean series.
# A series is basically a DataFrame with one column.
# print(df_pd["age"] == 30)

# Multiple conditions can be combined with & (and) or | (or).
# print(df_pd[(df_pd["age"] > 30) & (df_pd["age"] < 35)].head())
print(df_pd[(df_pd["age"] == 90) | (df_pd["native-country"] == " Hungary")])
````

**What to notice after the pandas filter:**
- The expression inside `df_pd[...]` is a Boolean Series: one `True` or `False` value for each row.
- pandas keeps the rows where the condition is `True`.
- Parentheses are required around each condition when combining conditions.


## Filter rows in Polars
### 1) Integer-based indexing
- There are no .loc[] or .iloc[] methods in Polars, but you can still select
  rows as you would with NumPy arrays.
- df_pl[start:stop:step]


````python
# Example:
df_pl[1:10:2]
````

**What to notice after the Polars integer-slice example:**
- Polars does not use pandas-style `.loc` or `.iloc`.
- You can still select rows by position using slice syntax such as `df_pl[1:10:2]`.
- For condition-based filtering, use `.filter()`.


### 2) Select rows based on column conditions
- the syntax is similar, except that you use .filter()

````python
df_pl.filter((df_pl["age"] == 30) & (df_pl["native-country"] == " India"))
````

**What to notice after the Polars filter:**
- The structure is similar to pandas, but the condition goes inside `.filter(...)`.
- The result should be interpreted the same way: only rows matching the condition are returned.
- Comparing the pandas and Polars versions helps separate the data idea from the package-specific syntax.
